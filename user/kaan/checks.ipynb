```python
from pathlib import Path
import sys

sys.path.insert(0, str(Path.cwd().parents[1]))


import os
import fit.Likelihood as L
```

```python
class DotDict(dict):
    """dot.notation access to dictionary attributes"""
    __getattr__ = dict.get
    __setattr__ = dict.__setitem__
    __delattr__ = dict.__delitem__

args = DotDict()
args.config = os.path.join(sys.path[0], "configs/unbinned_merged.yaml")
args.no_syst = True
```

```python
import common.yaml_loader as yaml_loader 
cfg = yaml_loader.load_yaml(args.config)
yaml_loader.print_summary(cfg, args.config, yaml_loader._INCLUDE_TRACE)
yaml_loader.load_surrogates(cfg, args.config, overwrite=False, prefer_numba=False)

like_info = L.load_likelihood(cfg)

hyp = L.build_hypothesis_from_likelihood(like_info, name="SR")

# --- optionally disable all nuisances ---
if args.no_syst:
    for p in hyp.nuisances:
        p.val = 0.0
        p.isFrozen = True
    print("[opts] --no_syst: all nuisances set to 0 and frozen.")

hyp.print()
```

```python
n2ll = L.N2LL( like_info, 'data.samples',  os.path.join( "NN2LCache",  os.path.splitext(os.path.basename(args.config))[0], cfg['version']), cache_root=None, overwrite=args.overwrite)
n2ll.build_cache()
n2ll.prepare_runtime()

# compute Asimov
n2ll.setAsimov()

# L1411: (B) ASIMOV MODE
```
